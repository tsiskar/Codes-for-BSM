```python
import sympy as sp

import numpy as np
from IPython.display import display, Math
from sympy import pprint, simplify, expand, sstr, latex
from matplotlib import pyplot as plt

def cos_to_sin2_halfangle_s(expr: sp.Expr) -> sp.Expr:
    """Replace cos(u) -> 1 - 2*sin(u/2)**2 everywhere (no phase shift)."""
    u = sp.Wild('u')
    out = expr.replace(sp.cos(u), 1 - 2*sp.sin(u/2)**2)
    expr = str(sp.expand(out))
    return (expr
    )

def pretty_s(expr):
    expr = str(expr)
    return (expr
        .replace("**6", "⁶")
        .replace("**5", "⁵")
        .replace("**4", "⁴")
        .replace("**3", "³")
        .replace("**2", "²")
        .replace("*", "")
    )

def sin_theta_from_tan2_all(tan2theta, theta):
    r = sp.sqrt(1 / (1 + tan2theta**2))

    sin_branches = [
        sp.sqrt((1 + r) / 2),
        -sp.sqrt((1 + r) / 2),
        sp.sqrt((1 - r) / 2),
        -sp.sqrt((1 - r) / 2),
    ]
    cos_branches = [
        sp.sqrt((1 - r)/2),
        -sp.sqrt((1 - r)/2),
        sp.sqrt((1 + r)/2),
        -sp.sqrt((1 + r)/2),
    ]

    #equations = [sp.Eq(sp.sin(theta), b) for b in branches]
    #equations = [sp.Eq(sp.sin(theta), sp.simplify(b)) for b in branches]

    for i, (s, c) in enumerate(zip(sin_branches, cos_branches), 1):
        print(f"\nBranch {i}:")
        sp.pprint(sp.Eq(sp.sin(theta), sp.simplify(s)))
        sp.pprint(sp.Eq(sp.cos(theta), sp.simplify(c)))

def heavy_simplify(e):
  e = sp.simplify(e)                  # light
  e = powsimp(e, force=True)          # combine powers aggressively
  e = sqrtdenest(e)                   # denest sqrt(sqrt(.)) when possible
  e = together(e
                )                     # combine rational parts
  e = cancel(e)                       # cancel common factors
  e = factor(e)                       # factor common stuff
  return e
def cos_to_sin2_halfangle(expr):
    return expr.replace(
        lambda e: e.func == sp.cos,
        lambda e: 1 - 2*sp.sin(e.args[0]/2)**2
    )
    return(sp.expand(expr))
```
